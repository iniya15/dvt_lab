```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Iniya\Downloads\Spotify_Youtube.csv")

print("Dataset loaded successfully")
```

```python
high_views = df[df["Views"] > 1000000]

print(high_views[["Artist", "Track", "Views"]].head(10))
```

```python
high_likes = df[df["Likes"] > 1000000]

print(high_likes[["Artist", "Track", "Likes"]].head(10))
```

```python
sorted_df = df.sort_values(
    by="Views",
    ascending=False
)

print(sorted_df[["Artist", "Track", "Views"]].head(10))
```

```python
top_songs = (
    df.groupby("Track")["Views"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 5))

top_songs.plot(kind="bar")

plt.title("Top 10 Songs by YouTube Views")
plt.xlabel("Song")
plt.ylabel("Total Views")
plt.xticks(rotation=45)

plt.show()
```
