```python
import pandas as pd

# Load dataset
df = pd.read_excel(r"C:\Users\Iniya\Downloads\tourism_dataset.xlsx")

# Create a copy for processing
tourism = df.copy()

# Display original dataset information
print("Original Shape:", df.shape)
print("\nColumn Names:")
print(tourism.columns.tolist())

print("\nMissing Values:")
print(tourism.isnull().sum())

# Remove duplicate records
tourism.drop_duplicates(inplace=True)

# Standardize column names
tourism.columns = (
    tourism.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

print("\nCleaned Shape:", tourism.shape)

# Display first 5 rows
tourism.head()
```

```python
from sqlalchemy import create_engine

username = "root"
password = "Iniya_15022006"
host = "localhost"
port = 3306
database = "tourism_analytics"

db_engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
)
```

```python
with db_engine.connect() as connection:
    print("Connected successfully!")
```

```python
tourism.to_sql(
    "tourism_data",
    con=db_engine,
    if_exists="replace",
    index=False
)
```
