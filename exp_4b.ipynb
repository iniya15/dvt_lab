```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Iniya\Downloads\Spotify_Youtube.csv")

print(df)
print(df.head())
print(df.shape)
print(df.dtypes)
print(df.describe())
print(df[df["Views"] > 1000000])
print(df[df["Likes"] > 10000])
print(df.sort_values("Views", ascending=False))
print(df.sort_values("Likes", ascending=False))
print(df.groupby("Artist")["Views"].mean().head(10))
print(df.groupby("Album_type")["Views"].mean())
print("Mean Views:", df["Views"].mean())
print("Median Views:", df["Views"].median())
print("Minimum Views:", df["Views"].min())
print("Maximum Views:", df["Views"].max())
print("Standard Deviation:", df["Views"].std())
print(df[["Views", "Likes", "Comments"]].corr())
print("Correlation:", df["Views"].corr(df["Likes"]))

top_artists = df.groupby("Artist")["Views"].sum().sort_values(ascending=False).head(10)
plt.bar(top_artists.index, top_artists.values)
plt.title("Top 10 Artists by Views")
plt.xlabel("Artist")
plt.ylabel("Total Views")
plt.xticks(rotation=45)
plt.show()

album_views = df.groupby("Album_type")["Views"].mean()
plt.bar(album_views.index, album_views.values)
plt.title("Average Views by Album Type")
plt.xlabel("Album Type")
plt.ylabel("Average Views")
plt.show()

plt.scatter(df["Views"], df["Likes"])
plt.title("Views vs Likes")
plt.xlabel("Views")
plt.ylabel("Likes")
plt.show()

plt.hist(df["Views"].dropna(), bins=10)
plt.title("Views Distribution")
plt.xlabel("Views")
plt.ylabel("Number of Songs")
plt.show()

plt.boxplot(df["Views"].dropna())
plt.title("Views Box Plot")
plt.ylabel("Views")
plt.show()
```
