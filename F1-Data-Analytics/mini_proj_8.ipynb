```python
import pandas as pd

# Load dataset
df = pd.read_csv(r"C:\Users\Iniya\Downloads\Database_Formula_1.csv")

# Create a copy for processing
f1 = df.copy()

# Display original dataset information
print("Original Shape:", df.shape)
print("\nColumn Names:")
print(f1.columns.tolist())

print("\nMissing Values:")
print(f1.isnull().sum())

# Remove duplicate records
f1.drop_duplicates(inplace=True)

# Standardize column names
f1.columns = (
    f1.columns
    .str.lower()
    .str.strip()
    .str.replace(" ", "_")
)

print("\nCleaned Shape:", f1.shape)

# Display first 5 rows
f1.head()
```
