```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\Iniya\Downloads\Spotify_Youtube.csv")
```

```python
# Line Chart
months = df["Track"].head(10)
sales = df["Views"].head(10)

plt.plot(months, sales)
plt.title("Views of Tracks")
plt.xlabel("Tracks")
plt.ylabel("Views")
plt.xticks(rotation=45)
plt.show()
```

```python
# Bar Chart
months = df["Track"].head(10)
sales = df["Views"].head(10)

plt.bar(months, sales)
plt.title("Views of Tracks")
plt.xlabel("Tracks")
plt.ylabel("Views")
plt.xticks(rotation=45)
plt.show()
```

```python
# Pie Chart
months = df["Track"].head(6)
sales = df["Views"].head(6)

plt.pie(sales, labels=months, autopct="%1.1f%%")
plt.title("Views Distribution")
plt.show()
```

```python
# Histogram
sales = df["Views"].dropna()

plt.hist(sales, bins=5)
plt.title("Views Distribution")
plt.xlabel("Views")
plt.ylabel("Frequency")
plt.show()
```

```python
# Scatter Plot
months = df["Track"].head(10)
sales = df["Views"].head(10)

plt.scatter(months, sales)
plt.title("Views of Tracks")
plt.xlabel("Tracks")
plt.ylabel("Views")
plt.xticks(rotation=45)
plt.show()
```

```python
# Box Plot
sales = df["Views"].dropna()

plt.boxplot(sales)
plt.title("Views Box Plot")
plt.ylabel("Views")
plt.show()
```

```python
# Multiple Line Charts
tracks = df["Track"].head(10)
views = df["Views"].head(10)
likes = df["Likes"].head(10)

plt.plot(tracks, views, label="Views")
plt.plot(tracks, likes, label="Likes")
plt.title("Views and Likes")
plt.xlabel("Tracks")
plt.ylabel("Value")
plt.xticks(rotation=45)
plt.legend()
plt.show()
```
