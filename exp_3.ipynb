```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\Iniya\Downloads\Spotify_Youtube.csv")

df.head()
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
df["Views"] = df["Views"].fillna(df["Views"].mean())
df["Likes"] = df["Likes"].fillna(df["Likes"].mean())
df["Comments"] = df["Comments"].fillna(df["Comments"].mean())
df["Stream"] = df["Stream"].fillna(df["Stream"].mean())

print(df.isnull().sum())
```

```python
print("Duplicate records:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

print("Duplicate records after removal:", df.duplicated().sum())
```

```python
Q1 = df["Views"].quantile(0.25)
Q3 = df["Views"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["Views"] < lower) |
    (df["Views"] > upper)
]

print(outliers[["Artist", "Track", "Views"]])
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df["Views_Normalized"] = scaler.fit_transform(df[["Views"]])

print(df[["Views", "Views_Normalized"]].head())
```
