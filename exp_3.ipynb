```python
import pandas as pd
df = pd.read_csv(r"C:\Users\Iniya\Downloads\Spotify_Youtube.csv")
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
df["Views"] = df["Views"].fillna(df["Views"].mean())
print(df)
```

```python
print(df.duplicated())
df.drop_duplicates(inplace=True)
print(df)
```

```python
Q1 = df["Views"].quantile(0.25)
Q3 = df["Views"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df["Views"] < lower) | (df["Views"] > upper)]
print(outliers)
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df[["Views"]] = scaler.fit_transform(df[["Views"]])
print(df[["Views"]])
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df["Album_type"] = encoder.fit_transform(df["Album_type"])
print(df)
```
